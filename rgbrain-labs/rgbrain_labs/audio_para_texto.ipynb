```python
engine = pyttsx4.init()
# rate
# Integer speech rate in words per minute. Defaults to 200 word per minute.

# voice
# String identifier of the active voice.

# voices
# List of pyttsx3.voice.Voice descriptor objects.

# volume
# Floating point volume in the range of 0.0 to 1.0 inclusive. Defaults to 1.0.

engine.setProperty('rate', 150) # Ajuste a velocidade
voices = engine.getProperty('voices')
engine.setProperty('voice', voices[53].id)  # Escolha a primeira voz disponível
engine.setProperty('volume', 0.5)
```

```python
# Text to speech list of amplitude values as output
text = """
O Consórcio União, um brilho de ouro,
Que atrai sonhadores, corações ávidos e louros.
Sem juros a brilhar, uma promessa sedutora,
Para realizar sonhos, sem a amarga tortura.
Parcelas suaves, como o vento que balança o pinheiro,
Um caminho sem espinhos, a um futuro mais inteiro.
"""
wav = tts.tts_to_file(
    text=text, 
    speaker_wav="audio_para_texto/exemplo_02.mp3", 
    language="pt", 
    file_path="audio_treinado_1.wav")
```
